SEM AVG is directly calculated from SEM 1 – 5, this creates data leakage. so we are only using marks from sem 1-3

```python
feature_cols = [
    'TOTAL MARKS OBTAINED IN ALL THE SUBJECTS IN CLASS X',
    'TOTAL MARKS OF EXAM APPEARED IN CLASS X',
    'TOTAL MARKS OBTAINED IN ALL THE SUBJECTS IN CLASS XII',
    'DIPLOMA AGGREGATE MARKS % ',
    'ENTRANCE RANK(OVERALL ALL INDIA RANK)',
    'SEM 1',
    'SEM 2',
    'SEM 3'
]

X = df[feature_cols].copy()
y = df['SEM AVG'].copy()
```

```python
import numpy as np
import re

X = X.replace(['N.A', '-', 'NA', 'na', ''], np.nan)
y = y.replace(['N.A', '-', 'NA', 'na', ''], np.nan)

def extract_number(x):
    if isinstance(x, str):
        m = re.search(r'(\d+(\.\d+)?)', x)
        return float(m.group(1)) if m else np.nan
    return x

X = X.applymap(extract_number)
y = y.apply(extract_number)
```

```python
mask = y.notna()
X = X.loc[mask]
y = y.loc[mask]
```

```python
import re

def extract_number(x):
    if isinstance(x, str):
        m = re.search(r'(\d+(\.\d+)?)', x)
        return float(m.group(1)) if m else None
    return x

X = X.applymap(extract_number)
```

```python
from sklearn.model_selection import KFold, cross_validate
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from lightgbm import LGBMRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import AdaBoostRegressor, HistGradientBoostingRegressor
from sklearn.linear_model import Ridge
```

```python
models = {
    "Random Forest": RandomForestRegressor(random_state=42),

    "Extra Trees": ExtraTreesRegressor(random_state=42),

    "XGBoost": XGBRegressor(
        random_state=42,
        verbosity=0
    ),

    "CatBoost": CatBoostRegressor(
        random_state=42,
        verbose=0
    ),



    "SVR": SVR(),

    "KNN": KNeighborsRegressor(),

    "AdaBoost": AdaBoostRegressor(random_state=42),

    "HistGB": HistGradientBoostingRegressor(random_state=42),

    "Ridge": Ridge()
}
```

```python
mask = y.notna()

X = X[mask]
y = y[mask]
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="mean")
X = imputer.fit_transform(X)
```

RMSE → average prediction error (strong penalty)

MAE → average simple error

MSE → squared error

R² → goodness of fit

MAPE → percentage error

```python
from sklearn.model_selection import cross_validate
import numpy as np

scoring = {
    'rmse': 'neg_root_mean_squared_error',
    'mae': 'neg_mean_absolute_error',
    'mse': 'neg_mean_squared_error',
    'r2': 'r2',
    'mape': 'neg_mean_absolute_percentage_error'
}

for name, model in models.items():

    scores = cross_validate(
        model,
        X,
        y,
        cv=kf,
        scoring=scoring
    )

    rmse = -scores['test_rmse'].mean()
    mae  = -scores['test_mae'].mean()
    mse  = -scores['test_mse'].mean()
    mape = -scores['test_mape'].mean()
    r2   = scores['test_r2'].mean()

    print(name)
    print("RMSE :", rmse)
    print("MAE  :", mae)
    print("MSE  :", mse)
    print("MAPE :", mape)
    print("R2   :", r2)
    print("------------------")
```

hypertunning best 5 models.


Hyperparameter tuning is the process of selecting the optimal model configuration parameters in order to improve predictive performance.
Trying different model settings and picking the one that gives the lowest RMSE.

```python
import optuna
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.ensemble import ExtraTreesRegressor

def objective_et(trial):

    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 400),
        "max_depth": trial.suggest_int("max_depth", 3, 25),
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 10),
        "random_state": 42
    }

    model = ExtraTreesRegressor(**params)

    scores = cross_val_score(
        model, X, y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    return -scores.mean()
```

```python
study_et = optuna.create_study(direction="minimize")
study_et.optimize(objective_et, n_trials=100)

print("Extra Trees best RMSE:", study_et.best_value)
print(study_et.best_params)
```

```python
from catboost import CatBoostRegressor

def objective_cat(trial):

    params = {
        "iterations": trial.suggest_int("iterations", 200, 600),
        "depth": trial.suggest_int("depth", 4, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.2),
        "random_state": 42,
        "verbose": 0
    }

    model = CatBoostRegressor(**params)

    scores = cross_val_score(
        model, X, y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    return -scores.mean()
```

```python
study_cat = optuna.create_study(direction="minimize")
study_cat.optimize(objective_cat, n_trials=100)

print("CatBoost best RMSE:", study_cat.best_value)
print(study_cat.best_params)
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

def objective_hgb(trial):

    params = {
        "max_iter": trial.suggest_int("max_iter", 100, 400),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.2),
        "max_depth": trial.suggest_int("max_depth", 3, 12),
        "random_state": 42
    }

    model = HistGradientBoostingRegressor(**params)

    scores = cross_val_score(
        model, X, y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    return -scores.mean()
```

```python
study_hgb = optuna.create_study(direction="minimize")
study_hgb.optimize(objective_hgb, n_trials=100)

print("HistGB best RMSE:", study_hgb.best_value)
print(study_hgb.best_params)
```

```python
from sklearn.ensemble import RandomForestRegressor

def objective_rf(trial):

    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 400),
        "max_depth": trial.suggest_int("max_depth", 3, 25),
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 10),
        "random_state": 42
    }

    model = RandomForestRegressor(**params)

    scores = cross_val_score(
        model, X, y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    return -scores.mean()
```

```python
study_rf = optuna.create_study(direction="minimize")
study_rf.optimize(objective_rf, n_trials=100)

print("Random Forest best RMSE:", study_rf.best_value)
print(study_rf.best_params)
```

```python
from xgboost import XGBRegressor

def objective_xgb(trial):

    params = {
        "n_estimators": trial.suggest_int("n_estimators", 100, 400),
        "max_depth": trial.suggest_int("max_depth", 3, 10),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.2),
        "subsample": trial.suggest_float("subsample", 0.7, 1.0),
        "random_state": 42,
        "verbosity": 0
    }

    model = XGBRegressor(**params)

    scores = cross_val_score(
        model, X, y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    return -scores.mean()
```

```python
study_xgb = optuna.create_study(direction="minimize")
study_xgb.optimize(objective_xgb, n_trials=100)

print("XGBoost best RMSE:", study_xgb.best_value)
print(study_xgb.best_params)
```
